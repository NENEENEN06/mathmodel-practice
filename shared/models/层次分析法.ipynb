```python
import numpy as np

A = np.array([[1,2,3],[1/2,1,2],[1/3,1/2,1]])
n = A.shape[0]
eig_val,eig_vec = np.linalg.eig(A)
max_val = np.max(eig_val)
CI = (max_val-n)/(n-1)
RI = [0,0.0001,0.52,0.89,1.12,1.26,1.36,1.41,1.46,1.49]
CR = CI/RI[n-1]
if(CR<0.1):
    print("一致性检验通过")
else:
    print("一致性检验不通过")
```

算数平均法

```python
import numpy as np

A = np.array([[1,2,3],[1/2,1,2],[1/3,1/2,1]])
ASum = np.sum(A,axis=0)
AAvg = A/ASum
W = np.mean(AAvg,axis=1)
print("权重向量为：",W)
```

几何平均法

```python
import numpy as np

A = np.array([[1,2,3],[1/2,1,2],[1/3,1/2,1]])
n = A.shape[0]
W = np.prod(A,axis=1)**(1/n)
W = W/np.sum(W)
print("权重向量为：",W)
```

特征值法

```python
import numpy as np

A = np.array([[1,2,3],[1/2,1,2],[1/3,1/2,1]])
eig_val,eig_vec = np.linalg.eig(A)
max_val = np.max(eig_val)
max_index = np.argmax(eig_val)
max_vec = eig_vec[:,max_index]
W = max_vec/np.sum(max_vec)
print("权重向量为：",W)
```
